```python
from datetime import datetime

def display_future_leap_years(final_year):
    current_year = datetime.now().year 
    
    if final_year < current_year:
        print(f"Please enter a year greater than or equal to {current_year}.")
        return
        
    print(f"\nLeap years between {current_year} and {final_year}:")
    for year in range(current_year, final_year + 1):
        if (year % 4 == 0 and year % 100 != 0) or (year % 400 == 0):
            print(year)
try:
    user_input = int(input("Enter the final year: "))
    display_future_leap_years(user_input)
except ValueError:
    print("Please enter a valid integer for the year.")
```

```python
#Create Rectangle class with attributes length and breadth and methods to find area and perimeter. Compare two Rectangle objects by their area.
class Rectangle:
    length = 10  
    breadth = 5  

    def get_area(self):
        return self.length * self.breadth

    def compare_area(self, other):
        if self.get_area() > other.get_area():
            print("First is larger")
        elif self.get_area() < other.get_area():
            print("Second is larger")
        else:
            print("Both are equal")
rect1 = Rectangle() 
rect2 = Rectangle() 

rect2.length = 8    
rect2.breadth = 7   

rect1.compare_area(rect2) 
```

```python
class BankAccount:
    # Explicit method acting as a manual initialization tool
    def create_account(self, account_no, name, account_type, balance=0.0):
        self.account_no = account_no
        self.name = name
        self.account_type = account_type
        self.balance = float(balance)

    def deposit(self, amount):
        if amount > 0:
            self.balance += amount
            print(f"₹{amount} deposited successfully.")
        else:
            print("Invalid deposit amount!")

    def withdraw(self, amount):
        if amount > self.balance:
            print("Transaction Denied: Insufficient balance!")
        elif amount <= 0:
            print("Invalid withdrawal amount!")
        else:
            self.balance -= amount
            print(f"₹{amount} withdrawn successfully.")

    def display_status(self):
        print("\n--- Account Summary ---")
        print(f"Account Number : {self.account_no}")
        print(f"Holder Name    : {self.name}")
        print(f"Account Type   : {self.account_type}")
        print(f"Current Balance: ₹{self.balance:.2f}")
        print("-----------------------\n")



acc1 = BankAccount()
acc1.create_account(1001, "Rahul Sharma", "Savings", 5000.0)

acc2 = BankAccount()
acc2.create_account(1002, "Priya Patel", "Current", 12000.0)

acc1.display_status()
acc1.deposit(1500)
acc1.display_status()
```

```python
class BankAccount:
    def __init__(self, name, balance):

        self.name = name
        self.balance = balance

    def deposit(self, amount):
        
        self.balance += amount
        print(f"Deposited: {amount}")

    def withdraw(self, amount):
        
        if amount > self.balance:
            print("Insufficient balance!")
        else:
            self.balance -= amount
            print(f"Withdrawn: {amount}")

    def display(self):
        
        print(f"{self.name}'s Balance: {self.balance}")


account = BankAccount("Amit", 5000)

account.display()       
account.deposit(2000)   
account.withdraw(1500)  
account.display()       
```

```python
class Time:
    def __init__(self, h, m, s):
        self.__h = h
        self.__m = m
        self.__s = s

    def __add__(self, other):
        tot_s = self.__s + other.__s
        tot_m = self.__m + other.__m + (tot_s // 60)
        tot_h = self.__h + other.__h + (tot_m // 60)
        
        
        return Time(tot_h, tot_m % 60, tot_s % 60)

    def display(self):
        # Format layout to look like 02:05:09
        print(f"{self.__h:02d}:{self.__m:02d}:{self.__s:02d}")


t1 = Time(2, 40, 50)
t2 = Time(1, 30, 20)

t3 = t1 + t2
t3.display()
```

```python

class Publisher:
    def __init__(self, name):
        self.name = name

    def display(self):
        print("Publisher:", self.name)

class Book(Publisher):
    def __init__(self, name, title, author):
        super().__init__(name)
        self.title = title
        self.author = author

    def display(self):
        super().display()  
        print("Title:", self.title)
        print("Author:", self.author)

class Python(Book):
    def __init__(self, name, title, author, price, pages):
        super().__init__(name, title, author)
        self.price = price
        self.pages = pages

    def display(self):
        super().display() 
        print("Price:", self.price)
        print("Pages:", self.pages)


my_book = Python("O'Reilly", "Learning Python", "Mark Lutz", 850, 600)
my_book.display()
```

```python
import csv

data = [
    {"Name": "Alice", "Age": 25, "City": "New York"},
    {"Name": "Bob", "Age": 30, "City": "Chicago"}
]

with open("data.csv", "w", newline="", encoding="utf-8") as f:
    writer = csv.DictWriter(f, fieldnames=data[0].keys())
    writer.writeheader()
    writer.writerows(data)

with open("data.csv", "r", newline="", encoding="utf-8") as f:
    for row in csv.DictReader(f):
        print(dict(row))
```
